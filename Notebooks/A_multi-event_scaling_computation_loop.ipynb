### Notebook A: Examine radial and transverse scaling for multiple families

Here we loop over several LFE families. For each family, we

- Read the detections and seismic data
- Group LFEs by time during the SSE
- Stack
- Compute the radial and transverse scaling between the early and late groups
- Save the results

```python
import numpy as np
import os
import matplotlib.pyplot as plt

import os,sys
sys.path.append(os.path.join(os.environ['PYFILES']))
import LFEAnalysis as lfeanalysis
lfeanalysis.reload_all(lfeanalysis)
```

### 1. Pick the families, region, and catalog to consider

```python
# note the region and catalog
region='Cascadia'
catalog='Bostock'

# and the data directory in case you need to change it
data_directory=None

# get the family numbers
lf=lfeanalysis.lfeanalyse(fnum=156,region='Cascadia',catalog='Bostock',data_directory=None)
lf.read_detection_info()
fnums=np.unique(lf.fnums)

# pick some families to analyse
# which families
#fnums=np.array([74,246,66,23,144,70,65,156,61,23,55,3,7,49,142,31,256,12,70,53,22,30,52,62,191,1,250],dtype=int)
#fnums=[74,156,250]
#fnums=np.unique(fnums)
fnums=fnums[fnums>113]
print('Families to analyse:\n',fnums)


```

### 2. And loop

```python
# time range to allow
import obspy
time_range=[obspy.UTCDateTime(2003,1,1),obspy.UTCDateTime(2005,1,1)]
time_range=[obspy.UTCDateTime(2005,1,1),obspy.UTCDateTime(2013,1,1)]
time_range=None

# loop over
for fnum in fnums:
    print('\nAnalysing scaling for family {:d}'.format(fnum))
    
    # as before, we need to initialize an analysis object
    lf=lfeanalysis.lfeanalyse(fnum=fnum,region=region,catalog=catalog,data_directory=data_directory)

    print('Reading and preparing the data')
    # and read the relevant detections
    lf.read_detection_info()

    # load the data
    lf.read_data()

    # normalise the seismograms
    lf.normalize_data(single_norm=False)

    # we may first need to discard any LFE intervals that contain NaNs, 
    # as these will disrupt the filtering
    lf.discard_data_with_nans()

    # and bandpass filter
    # we can filter the intervals with a 2-pass (non-causal) butterworth filter
    lf.filter_data(flm=[1,5])


    print('Grouping the LFEs by time')

    # identify the times of large SSEs
    lf.identify_sse_times(n_events=10)

    # here, compute the number of LFEs per 5-hour period
    lf.compute_lfe_rates(bin_duration=5*3600)

    # to do so, we first determine a family "start time": 
    # here the time when the LFE rate reaches half the maximum LFE rate
    lf.select_start_times(start_rate=0.5)

    # and then group the events by time since the start time
    # here we have two groups: 0-0.75 days since the start time and 0.75-3 days since the start time
    lf.group_events_bytime(tsplits=[0,0.75,3])


    print('Stacking')

    # stack for each group
    lf.stack_by_group(tlm=time_range)

    # and pick arrival times
    lf.pick_stacks(minsnr=10)


    print('Reading station information')

    # read the station locations
    lf.read_station_locations()

    # and compute the station location relative to the LFE family we're working with
    lf.relative_station_locations()

    # and then compute the takeoff angle per station
    lf.find_takeoff_angles()


    print('Computing scalings')

    # now we can compute scalings between the stacks
    lf.compute_scalings(wlen=[0,4])

    # let's go ahead and normalize the radial and transverse scalings for all stations
    lf.normalize_radtrans_scaling()


    print('Saving')

    # save the results
    lf.save_results()
    
```
