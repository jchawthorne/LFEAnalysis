```python
import numpy as np
import obspy
import lfeanalysis
import matplotlib.pyplot as plt
from importlib import reload
reload(lfeanalysis)
```

### 1. Loop over families and save the results

```python
# pick some families to analyse
# which families
fnums=np.array([74,246,66,23,144,70,65,156,61,23,55,3,7,49,142,31,256,12,70,53,22,30,52,62,191,1],dtype=int)
fnums=[74,156]
fnums=np.unique(fnums)
print(fnums)
```

```python
# loop over
for fnum in fnums:
    
    # as before, we need to initialize an analysis object
    lf=lfeanalysis.lfanalyse(fnum=fnum)

    # and load in the detections and waveforms
    lf.load_prep_data(flm=[1,8],single_norm=False)
    
    # stack for each group
    lf.stack_by_group()

    # and pick arrival times
    lf.pick_stacks(minsnr=10)
    
    # could think about adding amplitude weighting to improve the stacks
    
    # compute scalings
    lf.relative_station_locations()
    lf.compute_scalings(wlen=[0,4])
    
    # let's go ahead and normalize the radial and transverse scalings for all stations
    lf.normalize_radtrans_scaling()
    
    # pick some stations for duration classification
    lf.split_stations(prc_classify=0.75)
    
    # and estimate best-fitting amplitudes and durations
    olddurs=np.array([0.2,0.3])
    newdurs=np.arange(0.1,0.61,0.1)
    lf.duration_search(max_shift=0.2,olddurs=olddurs,newdurs=newdurs)
    
    # save the results
    lf.save_results()
    
```

```python
# note what results look like for one family
# one value per station
lf.scalingsc
```

### 7. Plot a histogram of the scaling differences

It is likely that the plot above just shows scatter due to noise, the scaling factors between these groups have significant noise compared to the values we'd expect from real signals.  So instead, we may wish to look at histograms of the determined scaling factor differences, or the mean or median of the scaling factor differences.  That turns out not to be very effective here, when we have only a few stations, but it will be more useful as we examine more LFEs and more stations.

```python
# there aren't many values here, but we can still have a start looking at the distributions
# the left column bins all the values, and the vertical line is the mean or median
# the right column recomputes the average multiple times, 
# bootstrapping by station or stack distribution, and plots the distribution of those values
lf.plot_binned_coefficients(chn='R',avetype='median')
```
